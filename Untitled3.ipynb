```python
import numpy as np
import matplotlib.pyplot as plt
import climlab
```

```python

# Partie I - Équilibre radiatif sans atmosphère

S0 = 1365
alpha = 0.3
epsilon = 1

sigma = climlab.constants.sigma

# Température d'équilibre
T = ((1 - alpha) * S0 / (4 * epsilon * sigma))**(1/4)

print("Température d'équilibre :", T, "K")
print("Température en Celsius :", T - 273.15, "°C")
```

```python

# Modèle avec une couche atmosphérique

def temperatures(S0, alpha, epsilon_a, epsilon_s):

    Ts = ((1 - alpha) * S0 /
          (4 * sigma * epsilon_s * (1 - epsilon_a / 2)))**(1/4)

    Ta = (epsilon_s * Ts**4 / 2)**(1/4)

    return Ta, Ts

# Paramètres de référence
S0 = 1365
alpha = 0.3
epsilon_a = 0.77
epsilon_s = 1

# Calcul des températures
Ta, Ts = temperatures(S0, alpha, epsilon_a, epsilon_s)

print("Température atmosphérique :", Ta, "K")
print("Température de surface :", Ts, "K")
```

```python

# Sensibilité à l'albédo planétaire

albedos = np.arange(0.1, 0.6, 0.1)

Ta_albedo = []
Ts_albedo = []

for alpha in albedos:

    Ta, Ts = temperatures(S0, alpha, epsilon_a, epsilon_s)

    Ta_albedo.append(Ta)
    Ts_albedo.append(Ts)

# Graphique
plt.plot(albedos, Ts_albedo, marker='o', label='Surface')
plt.plot(albedos, Ta_albedo, marker='o', label='Atmosphère')

plt.xlabel('Albédo planétaire')
plt.ylabel('Température (K)')
plt.title("Sensibilité des températures à l'albédo")
plt.legend()
plt.grid()
plt.savefig('sensibilite_albedo.png', dpi=300, bbox_inches='tight')
plt.show()

```

```python

# Sensibilité à l'émissivité atmosphérique

alpha = 0.3

emissivites = np.arange(0.1, 1.0, 0.1)

Ta_emissivite = []
Ts_emissivite = []

for epsilon_a in emissivites:

    Ta, Ts = temperatures(S0, alpha, epsilon_a, epsilon_s)

    Ta_emissivite.append(Ta)
    Ts_emissivite.append(Ts)

# Graphique
plt.plot(emissivites, Ts_emissivite, marker='o', label='Surface')
plt.plot(emissivites, Ta_emissivite, marker='o', label='Atmosphère')

plt.xlabel('Émissivité atmosphérique')
plt.ylabel('Température (K)')
plt.title("Sensibilité des températures à l'émissivité")
plt.legend()
plt.grid()

plt.savefig('sensibilite_emissivite.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python

# Partie II - Modèle 1-D d'équilibre radiatif

# Paramètres du modèle
alb = 0.25
Nz = 30

# Création de la colonne atmosphérique
state = climlab.column_state(num_lev=Nz)

# Humidité relative fixe
h2o = climlab.radiation.ManabeWaterVapor(
    name='WaterVapor', state=state)

# Rayonnement
rad = climlab.radiation.RRTMG(
    name='Radiation', state=state,
    specific_humidity=h2o.q, albedo=alb)

# Couplage du rayonnement et de la vapeur d'eau
rcm = climlab.couple([rad, h2o])

# Température initiale identique à tous les niveaux
rcm.state.Tatm[:] = rcm.state.Ts

print(state)
```

```python
# Avancer le modèle d'un pas de temps
rcm.step_forward()

# Afficher les flux radiatifs
print("ASR :", rcm.ASR, "W/m²")
print("OLR :", rcm.OLR, "W/m²")
```

```python

# Simulation de contrôle - Équilibre radiatif

# Faire évoluer le modèle pendant deux ans
rcm.integrate_years(2)

# Afficher les résultats
print("ASR :", rcm.ASR, "W/m²")
print("OLR :", rcm.OLR, "W/m²")

# Différence entre les flux
print("Différence :", rcm.ASR - rcm.OLR, "W/m²")

# Température de surface
print("Température de surface :", rcm.Ts, "K")
```

```python

# Partie II - Sensibilité au doublement du CO2

# Concentration initiale du CO2
co2_ref = rcm.absorber_vmr['CO2']

# Nouveau modèle
state_co2 = climlab.column_state(num_lev=30)

h2o_co2 = climlab.radiation.ManabeWaterVapor(
    name='WaterVapor', state=state_co2)

rad_co2 = climlab.radiation.RRTMG(
    name='Radiation', state=state_co2,
    specific_humidity=h2o_co2.q, albedo=0.25)

rcm_co2 = climlab.couple([rad_co2, h2o_co2])

# Température initiale
rcm_co2.state.Tatm[:] = rcm_co2.state.Ts

# Doublement de la concentration de CO2
rcm_co2.absorber_vmr['CO2'] = 2 * co2_ref

# Simulation de deux ans
rcm_co2.integrate_years(2)

# Comparaison des températures
print("Température de contrôle :", rcm.Ts, "K")
print("Température avec CO2 doublé :", rcm_co2.Ts, "K")
print("Différence :", rcm_co2.Ts - rcm.Ts, "K")
```

```python

# Partie II - Section 2.3
# Sensibilité à l'albédo de surface

albedos_surface = [0.15, 0.25, 0.35]

Ts_surface = []
modeles_albedo = []

for a in albedos_surface:

    # Nouveau modèle
    state_a = climlab.column_state(num_lev=30)

    h2o_a = climlab.radiation.ManabeWaterVapor(
        name='WaterVapor', state=state_a)

    rad_a = climlab.radiation.RRTMG(
        name='Radiation', state=state_a,
        specific_humidity=h2o_a.q, albedo=a)

    rcm_a = climlab.couple([rad_a, h2o_a])

    # Température initiale
    rcm_a.state.Tatm[:] = rcm_a.state.Ts

    # Simulation de deux ans
    rcm_a.integrate_years(2)

    # Conservation des résultats
    Ts_surface.append(rcm_a.Ts[0])
    modeles_albedo.append(rcm_a)

    print("Albédo :", a)
    print("Température de surface :", rcm_a.Ts, "K")
    print("ASR - OLR :", rcm_a.ASR - rcm_a.OLR, "W/m²")
```

```python

# Partie III - Modèle radiatif-convectif

# Création de la colonne atmosphérique
state_conv = climlab.column_state(num_lev=30)

# Humidité relative fixe
h2o_conv = climlab.radiation.ManabeWaterVapor(
    name='WaterVapor', state=state_conv)

# Rayonnement
rad_conv = climlab.radiation.RRTMG(
    name='Radiation', state=state_conv,
    specific_humidity=h2o_conv.q, albedo=0.25)

# Ajustement convectif
conv = climlab.convection.ConvectiveAdjustment(
    name='Convection', state=state_conv,
    adj_lapse_rate=6.5)

# Couplage des processus
rcm_conv = climlab.couple([rad_conv, conv, h2o_conv])

# Température initiale
rcm_conv.state.Tatm[:] = rcm_conv.state.Ts

# Simulation de deux ans
rcm_conv.integrate_years(2)

# Résultats
print("Température sans convection :", rcm.Ts, "K")
print("Température avec convection :", rcm_conv.Ts, "K")

print("ASR - OLR :",
      rcm_conv.ASR - rcm_conv.OLR, "W/m²")
```
